# Grouping and Aggregation of data 

```python
import pandas as pd
```

```python
df=pd.read_csv("raw_data.csv")
```

```python
df.groupby("country")["income"].mean()
df.groupby("country")["income"].sum()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()
```

```python
df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].sum()
df.groupby("gender")["income"].min()
df.groupby("gender")["income"].max()
```

```python
### aggregate  function 
df.groupby("country")["income"].agg(["mean","sum","min","max"])
df.groupby("country")["income"].aggregate(avg_salary="mean",sum_salary="sum", min_salary="min", max_salary="max")
```

```python
df.groupby("country").agg({
    "income" : "mean",
    "age" : "mean"
})
```

```python
df.groupby("country").agg(
    avg_salary=("income","mean"),
    avg_age=("age","mean")
)
```
