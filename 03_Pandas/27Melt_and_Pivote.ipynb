# Melt and pivote functions of pandas

```python
import pandas as pd
```

```python
df=pd.read_csv("raw_data.csv")
```

```python
df
```

### Melt function => use to convert Wide to long formate

```python
melted_df=df.melt(
    id_vars=["id", "name", "country", "gender"],
    value_vars=["age", "income"],
    var_name="variable",
    value_name="value"
)
melted_df
```

### Pivote function => use to convert Long to Wide formate

```python
# Remove dupicates before applying pivote 

melted_df=melted_df.drop_duplicates(
    subset=["id", "name", "country", "gender", "variable"]
)

melted_df.duplicated(
    subset=["id", "name", "country", "gender", "variable"]
).sum()

pivoted_df = melted_df.pivot(
    index=["id", "name", "country", "gender"],
    columns="variable",
    values="value"
)
pivoted_df
```
