# Merge and join data

```python
import pandas as pd
```

```python
customer_df=pd.DataFrame(
    {
        "customer_id":[1,2,3,4],
        "name":["Adam", "Bob","Charlie", "Dave"]
    }
)
```

```python
order_df=pd.DataFrame({
    "order_id":[101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})
```

```python
# inner join (in this the data common in both table will be add to the new table)
pd.merge(customer_df, order_df, on="customer_id", how="inner")  # default inner
```

```python
# left join (in this the data common and all data in first table will be added to the new table)
pd.merge(customer_df, order_df, on="customer_id" ,how="left")
```

```python
# right join (in this the data common and all data in second table will be added to the new table)
pd.merge(customer_df, order_df, on="customer_id" ,how="right")
```

```python
# outer join (in this the uinon of both the table is sored in the new table)
pd.merge(customer_df, order_df, on="customer_id" ,how="outer")
```
