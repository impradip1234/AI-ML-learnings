```python
import pandas as pd
```

```python
#homogeneous ---> series stores data in same data type........
s=pd.Series(["pradip", "aditya", "satish", "ayush", 123], index=[1,2,3,4,5])
print(s)
```

```python
'''Vectorized operations -->Vectorized operations are operations
that performa calculation on multiple elements of a data structure 
(such as an array #or vector) at the same time, instead of processing each
element individually using a loop'''
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])
print(s1+s2)
```

```python
# Handle missing values with NaN
s1[0]=100
print(s1)
```

```python
# Drop function for removing value in series
changed_s1=s1.drop(0)
print(s1)
print(changed_s1)
```
