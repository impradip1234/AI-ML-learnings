```python
import numpy as np
```

```python
# Aggregation.....
arr=np.array([1,2,3,4,5])
print(np.sum(arr))
print(np.prod(arr))
print(np.max(arr))
print(np.min(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.var(arr))
print(np.argmin(arr))
print(np.argmax(arr))
```

```python
# Power

arr=np.array([1,2,3,4])
print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr,3))
print(np.pow(2,4))
```

```python
# Rounding
a=3.4455
print(np.round(3.14))
print(np.round(arr))
print(np.round(a))
print(np.ceil(a))
print(np.floor(a))
print(np.trunc(a))
```

```python
# Log and Expontential
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
```

```python
# Extra....
print(np.unique(arr))
print(np.unique([2,2,2,23,3,3,4]))
print(np.sort(arr))
print(np.abs(arr))
print(np.abs(-12))
```
