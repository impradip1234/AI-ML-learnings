# Thing in terms of data

### Data loading

```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
    return data
data = load_data("2store_data.json")
print(data)
print(type(data))
```

### Data cleaning

```python
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_user=set()

    for user in data:
        #cleaning ratings
        raw_rating=str(user["rating"]).strip().lower()
        
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
            
        user["rating"] = raw_rating

        #handle missing values
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        #deduplication
        if(user["name"].strip() in unique_user):
            continue
        else:
            unique_user.add(user["name"])
            cleaned_data.append(user)
    return cleaned_data
```

```python
store_data = load_data("2store_data.json")
final_data = clean_data(store_data)
print(final_data)
```

### Geting meaningfull insights from data

```python
#average rating
def avg_rating(data):
    total_rating=0
    for user in data:
        total_rating+=float(user["rating"])
    print(f"average rating is:{total_rating/len(data)}")

#percentage of users with poor rating 
def percentage_poor_rating(data):
    total_rating=len(data)
    poor_rating=0
    for user in data:
        if(float(user["rating"])<3):
            poor_rating+=1
    print(f"percentage of poor rating = {(poor_rating/total_rating)*100}%")
```

```python
avg_rating(final_data)
percentage_poor_rating(final_data)
```

### Building recomendation system

```python
def get_recommendations(data):
    recommendation=[]

    for user in data:
        cur_recomm={}
        cur_recomm["name"]=user["name"]
        if float(user["rating"])>=4:
            cur_recomm["brand"]="Apple"
        else:
            cur_recomm["brand"]="Samsung"

        recommendation.append(cur_recomm)
    return recommendation
```

```python
get_recommendations(final_data)
```
