```python
import numpy as np
```

```python
import time
```

```python
size=10_000_000
```

```python
#python list
py_list=list(range(size))
start=time.time()
sq_list=[x**2 for x in py_list]
end=time.time()
print(f"py_list time is : {end-start}")

np_arr=np.array(py_list)
start=time.time()
sq_arr=np_arr**2
end=time.time()
print(f"np_arr time is :{end-start}")
```

# Array takes lesser memory size than python list

```python
import sys
```

```python
print(f"size of the py_list={sys.getsizeof(py_list)}")
print(f"size of the np_arr ={np_arr.nbytes}")
```
