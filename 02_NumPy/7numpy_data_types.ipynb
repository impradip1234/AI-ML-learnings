```python
import numpy as np
```

```python
# Data Types
#int32 and int64 => system dependent
arr=np.array([22,4,43,5443,3])
print(arr.dtype)
```

```python
#floating point nums: float32 and float64
arr1=np.array([3.3,2.4,233.3,0.2])
print(arr1.dtype)
```

```python
#boolean: bool
arr2=np.array([True,False,False,False,True])
print(arr2.dtype)
```

```python
# complex nums: complex64 and complex128
arr3=np.array([5+4j])
arr4=np.array([5+4j])
print(arr3+arr4)
print(arr3.dtype)
```

```python
#string : S(bytes_str) and U(unicode_str)
arr5=np.array(['a','b'])
print(arr5.dtype)
```

```python
#Object : generic python objects = object
arr6=np.array(["pradip",'a',{2,3,4},44,4.4])
print(arr6.dtype)
```
