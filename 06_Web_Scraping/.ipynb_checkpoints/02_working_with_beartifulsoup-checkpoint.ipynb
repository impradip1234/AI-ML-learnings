```python
# for h3 in all_h3:
#     print(h3.attrs)
```

```python
all_countries = []

for h3 in all_h3:
    country_name = h3.get_text(strip=True)
    
    population = h3.find_next("div").select_one(
        "span.country-population"
    ).get_text(strip=True)
    
    all_countries.append([country_name, population])
```

```python
# all_countries
import pandas as pd
df=pd.DataFrame(all_countries)
```
